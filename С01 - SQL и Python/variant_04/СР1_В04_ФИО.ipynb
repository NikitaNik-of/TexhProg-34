### Задание 4. Работа с SQL в SQLite

#### 4.1. Создание таблиц и импорт данных
В коде ниже таблицы `bar_orders`, `hardware_pcs`, `tournaments` уже сформированы в качестве примера. Допишите DDL-скрипт для оставшихся таблиц:

**Таблица `members`**:  
Таблица содержит целочисленный первичный ключ `member_id` и обязательное текстовое поле `nickname`.  
Поля `email`, `city` и `registered_at` хранятся в виде текста.  
Поле `loyalty_tier` принимает только текстовые значения 'Bronze', 'Silver', 'Gold' или 'Platinum'.  
Баланс `balance` задается вещественным числом со значением не меньше нуля.

**Таблица `game_sessions`**:  
Таблица содержит целочисленный первичный ключ `session_id`.  
Поля `member_id` и `pc_id` являются внешними ключами со ссылками на таблицы `members` и `hardware_pcs`.  
Поля `game_title` и `session_date` являются обязательными текстовыми полями.  
Длительность `duration_hours` задается положительным вещественным числом.  
Признак промокода `discount_promo` ограничен целыми значениями 0 или 1.  
Сумма оплаты `total_paid` принимает неотрицательные вещественные значения, а рейтинг `client_rating` хранится в виде вещественного числа.

После завершения описания схемы выполните ячейку для создания таблиц, импорта CSV-файлов и сверки количества записей.

````python
import sqlite3
import pandas as pd
import pathlib
import os

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()
cursor.execute("PRAGMA foreign_keys = ON;")

ddl_script = '''
CREATE TABLE members (
    member_id ...,
    nickname ...,
    email ...,
    city ...,
    registered_at ...,
    loyalty_tier ...,
    balance ...
);

-- Пример
CREATE TABLE hardware_pcs (
    pc_id       INTEGER PRIMARY KEY,
    pc_number   TEXT NOT NULL UNIQUE,
    zone_name   TEXT NOT NULL,
    hourly_rate REAL CHECK (hourly_rate > 0),
    gpu         TEXT,
    cpu         TEXT,
    ram_gb      INTEGER CHECK (ram_gb > 0),
    status      TEXT CHECK (status IN ('Active', 'Maintenance', 'Reserved'))
);

CREATE TABLE game_sessions (
    session_id ...,
    member_id ...,
    pc_id ...,
    game_title ...,
    session_date ...,
    duration_hours ...,
    discount_promo ...,
    total_paid ...,
    client_rating ...
);

-- Пример
CREATE TABLE bar_orders (
    order_id     INTEGER PRIMARY KEY,
    member_id    INTEGER REFERENCES members(member_id),
    item_name    TEXT NOT NULL,
    category     TEXT NOT NULL,
    order_date   TEXT NOT NULL,
    quantity     INTEGER CHECK (quantity > 0),
    item_price   REAL CHECK (item_price > 0),
    total_amount REAL CHECK (total_amount >= 0)
);

-- Пример
CREATE TABLE tournaments (
    tournament_id   INTEGER PRIMARY KEY,
    tournament_name TEXT NOT NULL,
    game_title      TEXT NOT NULL,
    prize_fund      REAL CHECK (prize_fund >= 0),
    entry_fee       REAL CHECK (entry_fee >= 0),
    start_date      TEXT NOT NULL,
    min_rank        TEXT
);
'''

# 1. Создание таблиц
cursor.executescript(ddl_script)

# 2. Загрузка данных из CSV
data_dir = pathlib.Path("members.csv").parent if os.path.exists("members.csv") else pathlib.Path(f"../variant_04")
for table in ["members", "hardware_pcs", "game_sessions", "bar_orders", "tournaments"]:
    df_temp = pd.read_csv(data_dir / f"{table}.csv", sep=";")
    df_temp.to_sql(table, conn, if_exists="append", index=False)

# 3. Проверка количества строк
check_query = '''
SELECT 'members' AS tbl, COUNT(*) AS cnt FROM members
UNION ALL SELECT 'hardware_pcs', COUNT(*) FROM hardware_pcs
UNION ALL SELECT 'game_sessions', COUNT(*) FROM game_sessions
UNION ALL SELECT 'bar_orders', COUNT(*) FROM bar_orders
UNION ALL SELECT 'tournaments', COUNT(*) FROM tournaments;
'''
display(pd.read_sql_query(check_query, conn))
````
